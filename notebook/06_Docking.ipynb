```python
import pandas as pd
from rdkit import Chem
from rdkit.Chem import AllChem
```

```python
# Ordenar por la mejor predicción (mayor pChEMBL) y aislar el Top 5
df_elite = pd.read_csv('../results/CMNPD_Representantes_Elite.csv')
top_5 = df_elite.sort_values(by='Media_Predicciones', ascending=False).head(5)
print("Generando las estructuras 3D optimizadas de tus 5 mejores compuestos...")
```

```python
# 3. Iterar y guardar cada compuesto en un archivo individual
for index, row in top_5.reset_index().iterrows():
    smiles = row['Smiles']
    puntuacion = row['Media_Predicciones']
    mol = Chem.MolFromSmiles(smiles)
    
    if mol is not None:
        # Preparación 3D: Añadir hidrógenos, geometría y minimizar energía
        mol_h = Chem.AddHs(mol)
        AllChem.EmbedMolecule(mol_h, AllChem.ETKDG())
        AllChem.MMFFOptimizeMolecule(mol_h, maxIters=200)
        
        # Guardar en un archivo SDF individual (ej: Top_1_Pred_8.45.sdf)
        nombre_archivo = f'../results/Top_{index + 1}_Pred_{puntuacion:.2f}.sdf'
        writer = Chem.SDWriter(nombre_archivo)
        writer.write(mol_h)
        writer.close()
        print(f"Guardado: {nombre_archivo} (SMILES: {smiles[:15]}...)")

print("\n¡Proceso completado! Archivos listos para UCSF Chimera.")
```

```python
# 1. Extraer la fila completa del Top 3 (posición 2 en base 0)
fila_top3 = top_5.reset_index().iloc[2]

# 2. Imprimir todos los campos disponibles (identificador, smiles, etc.)
print("--- DATOS COMPLETOS DE TOP 3 ---")
for col, val in fila_top3.items():
  print(f"{col}: {val}")

# 3. Guardar el SMILES explícito para consultas
smiles_top3 = fila_top3['Smiles']
print(f"\nSMILES canónico de Top 3:\n{smiles_top3}")
```
